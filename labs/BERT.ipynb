```python
correct = 0
total = 0

with torch.no_grad():

    for batch in val_loader:
        input_ids = batch[0]
        attention_mask = batch[1]
        labels = batch[2]

        outputs = model(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

        predictions = torch.argmax(outputs.logits, dim=1)

        correct += (predictions == labels).sum().item()
        total += labels.size(0)

accuracy = correct / total

print("Validation Accuracy:", accuracy)
```

```python
def should_respond(text, threshold=0.80):
    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True
    )

    model.eval()

    with torch.no_grad():
        outputs = model(**inputs)
        probabilities = torch.softmax(outputs.logits, dim=1)

    probability = probabilities[0, 1].item()

    return probability >= threshold, probability
```

```python
texts = [
    "what is webkit",
    "I'm just reading about webkit",
    "can you explain this error",
    "I think I understand it now",
    "okay let me try this myself"
]

for text in texts:
    respond, probability = should_respond(text)

    print(
        text,
        "=>",
        "RESPOND" if respond else "SILENT",
        f"({probability:.3f})"
    )
```

```python
def should_respond(text, eagerness=0.5):
    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True
    )

    model.eval()

    with torch.no_grad():
        outputs = model(**inputs)
        probabilities = torch.softmax(outputs.logits, dim=1)

    probability = probabilities[0, 1].item()

    threshold = 0.95 - (eagerness * 0.50)

    return probability >= threshold, probability, threshold
```

```python
texts = [
    "what is webkit",
    "I'm just reading about webkit",
    "can you explain this error",
    "I think I understand it now",
    "okay let me try this myself"
]

for text in texts:
    respond, probability, threshold = should_respond(text)

    print(
        text,
        "=>",
        "RESPOND" if respond else "SILENT",
        f"({probability:.3f})",
        f"({threshold:.3f})"
    )
```
